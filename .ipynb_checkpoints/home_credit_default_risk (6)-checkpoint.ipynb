```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import PolynomialFeatures

poly_features = df[['EXT_SOURCE_1',
                    'EXT_SOURCE_2',
                    'EXT_SOURCE_3',
                    'DAYS_BIRTH']].copy()


poly_target = df['TARGET']

imputer = SimpleImputer(strategy='median')

poly_features = imputer.fit_transform(poly_features)

poly = PolynomialFeatures(degree=3)

poly_features = poly.fit_transform(poly_features)

feature_names = poly.get_feature_names_out(
    ['EXT_SOURCE_1',
     'EXT_SOURCE_2',
     'EXT_SOURCE_3',
     'DAYS_BIRTH']
)

poly_features = pd.DataFrame(poly_features, columns=feature_names)

poly_features['TARGET'] = poly_target.values

poly_corrs = poly_features.corr()['TARGET'].sort_values()

print("Most Negative Correlations")
print(poly_corrs.head(10))

print("\nMost Positive Correlations")
print(poly_corrs.tail(10))

poly_features = poly_features.drop(columns=['TARGET'])

poly_features['SK_ID_CURR'] = df['SK_ID_CURR'].values

df_poly = df.merge(poly_features, on='SK_ID_CURR', how='left')

print("Training data shape:", df_poly.shape)
```

```python
poly_features['SK_ID_CURR'] = df['SK_ID_CURR'].values

df_poly = df.merge(poly_features, on='SK_ID_CURR', how='left')

print("Training data shape:", df_poly.shape)
```

```python
train = df.drop(columns=['TARGET'])
train_labels = df['TARGET']

features = train.columns

imputer = SimpleImputer(strategy='median')
scaler = MinMaxScaler()

train = imputer.fit_transform(train)

train = scaler.fit_transform(train)


print("Training Shape:", train.shape)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
```

```python
X_new = df_new.drop('TARGET', axis=1)
y_new = df_new['TARGET']

X_train_new, X_test_new, y_train_new, y_test_new = train_test_split(
    X_new, y_new,
    test_size=0.2,
    random_state=42
)
```

```python
bool_cols=X_train_new.select_dtypes(include='bool').columns
num_cols=X_train_new.select_dtypes(include=['int64','float64']).columns
```

```python
X_train_new[bool_cols] = X_train_new[bool_cols].astype('Int64')
X_test_new[bool_cols] = X_test_new[bool_cols].astype('Int64')
```

```python
bool_imputer = SimpleImputer(strategy='most_frequent')
```

```python
X_train_new[bool_cols] = bool_imputer.fit_transform(
    X_train_new[bool_cols]
)
X_test_new[bool_cols] = bool_imputer.transform(
    X_test_new[bool_cols]
)
```

```python
X_train_new[bool_cols] = X_train_new[bool_cols].astype(int)
X_test_new[bool_cols] = X_test_new[bool_cols].astype(int)
```

```python
num_imputer = SimpleImputer(strategy='median')
```

```python
X_train_new[num_cols] = num_imputer.fit_transform(
    X_train_new[num_cols]
)

X_test_new[num_cols] = num_imputer.transform(
    X_test_new[num_cols]
)
```

```python
scaler = MinMaxScaler()
```

```python
X_train_new[num_cols] = scaler.fit_transform(
    X_train_new[num_cols]
)

X_test_new[num_cols] = scaler.transform(
    X_test_new[num_cols]
)
```

```python
print("X_train shape:", X_train_new.shape)
print("X_test shape:", X_test_new.shape)
print("y_train shape:", y_train_new.shape)
print("y_test shape:", y_test_new.shape)

print("\nRemaining NaN in X_train:",
      X_train_new.isna().sum().sum())

print("Remaining NaN in X_test:",
      X_test_new.isna().sum().sum())

print("\nData types:")
print(X_train_new.dtypes)
```

```python
from sklearn.metrics import roc_auc_score
```

```python
model = lgb.LGBMClassifier(
    n_estimators=10000,
    objective='binary',
    class_weight='balanced',
    learning_rate=0.05,
    reg_alpha=0.1,
    reg_lambda=0.1,
    subsample=0.8,
    n_jobs=-1,
    random_state=42
)
```

```python
X_train_new.columns = X_train_new.columns.astype(str).str.replace(
    r'[^A-Za-z0-9_]', '_', regex=True
)

X_test_new.columns = X_train_new.columns
```

```python
model.fit(
    X_train_new,
    y_train_new,
    eval_set=[
        (X_train_new, y_train_new),
        (X_test_new, y_test_new)
    ],
    eval_names=['train', 'test'],
    eval_metric='auc',
    callbacks=[
        lgb.early_stopping(100),
        lgb.log_evaluation(200)
    ]
)
```

```python
y_pred_new = model.predict_proba(X_test_new)[:, 1]
auc_new = roc_auc_score(y_test_new, y_pred_new)
```

```python
print("ROC-AUC:", auc_new)
```

```python
import matplotlib.pyplot as plt

results = model.evals_result_

plt.figure(figsize=(10, 6))

plt.plot(results['train']['auc'], label='Train AUC')
plt.plot(results['test']['auc'], label='Validation AUC')

plt.xlabel('Boosting Iterations')
plt.ylabel('ROC-AUC')
plt.title('LightGBM Training vs Validation AUC')
plt.legend()
plt.grid()

plt.show()
```

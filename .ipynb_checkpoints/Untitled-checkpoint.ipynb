```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
pd.set_option('display.max_columns',None)
```

```python
df_train = pd.read_csv("application_train.csv.zip")
```

```python
df_test = pd.read_csv("application_test.csv.zip")
```

```python
df_train.head()
```

```python
df_train.shape
```

```python
df_train.TARGET.value_counts()
```

```python
df_train.select_dtypes('object').apply(pd.Series.nunique,axis=0)
```

```python
df_train = df_train.drop(df_train[df_train['CODE_GENDER'] == 'XNA'].index)
```

```python
df_train.CODE_GENDER.value_counts()
```

```python
df_test.CODE_GENDER.value_counts()
```

```python
df_test.select_dtypes('object').apply(pd.Series.nunique,axis=0)
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
le = LabelEncoder()
```

```python
for col in df_train:
    if df_train[col].dtype =='object':
        if len(list(df_train[col].unique()))<=2:
            le.fit(df_train[col])
            df_train[col]=le.transform(df_train[col])
            df_test[col]=le.transform(df_test[col])
print("Done")
```

```python
df_train.head()
```

```python
df_test.head()
```

```python
df_train.select_dtypes('object').apply(pd.Series.nunique,axis=0)
```

```python
df_test.select_dtypes('object').apply(pd.Series.nunique,axis=0)
```

```python
df_train = pd.get_dummies(df_train)
df_test = pd.get_dummies(df_test)
```

```python
print('Training Features shape: ', df_train.shape)
print('Testing Features shape: ', df_test.shape)
```

```python
labels = df_train['TARGET']

df_train, df_test = df_train.align(df_test, join = 'inner', axis = 1)

df_train['TARGET'] = labels

print('Training Features shape: ', df_train.shape)
print('Testing Features shape: ', df_test.shape)
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import PolynomialFeatures

poly_features = df[['EXT_SOURCE_1',
                    'EXT_SOURCE_2',
                    'EXT_SOURCE_3',
                    'DAYS_BIRTH']].copy()


poly_target = df['TARGET']

imputer = SimpleImputer(strategy='median')

poly_features = imputer.fit_transform(poly_features)

poly = PolynomialFeatures(degree=3)

poly_features = poly.fit_transform(poly_features)

feature_names = poly.get_feature_names_out(
    ['EXT_SOURCE_1',
     'EXT_SOURCE_2',
     'EXT_SOURCE_3',
     'DAYS_BIRTH']
)

poly_features = pd.DataFrame(poly_features, columns=feature_names)

poly_features['TARGET'] = poly_target.values

poly_corrs = poly_features.corr()['TARGET'].sort_values()

print("Most Negative Correlations")
print(poly_corrs.head(10))

print("\nMost Positive Correlations")
print(poly_corrs.tail(10))

poly_features = poly_features.drop(columns=['TARGET'])

poly_features['SK_ID_CURR'] = df['SK_ID_CURR'].values

df_poly = df.merge(poly_features, on='SK_ID_CURR', how='left')

print("Training data shape:", df_poly.shape)
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
train = df.drop(columns=['TARGET'])
train_labels = df['TARGET']

features = train.columns

imputer = SimpleImputer(strategy='median')
scaler = MinMaxScaler()

train = imputer.fit_transform(train)

train = scaler.fit_transform(train)


print("Training Shape:", train.shape)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.model_selection import train_test_split

X_train, X_valid, y_train, y_valid = train_test_split(
    train,
    train_labels,
    test_size=0.2,
    random_state=42
)
```

```python
log_reg=LogisticRegression(C=0.0001)
```

```python
log_reg.fit(X_train, y_train)
```

```python
pred = log_reg.predict_proba(X_valid)[:,1]
```

```python
from sklearn.metrics import roc_auc_score
score = roc_auc_score(y_valid, pred)
print(score)
```

```python
from sklearn.ensemble import RandomForestClassifier
```

```python
random_forest = RandomForestClassifier(
    n_estimators=100,
    random_state=50,
    n_jobs=-1,
    verbose=1
)

random_forest.fit(X_train, y_train)

rf_pred = random_forest.predict_proba(X_valid)[:, 1]

rf_auc = roc_auc_score(y_valid, rf_pred)

print("Random Forest ROC-AUC:", rf_auc)
```
